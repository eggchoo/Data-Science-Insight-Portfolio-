# 淘宝用户购买行为数据分析

## 项目介绍

本项目基于淘宝用户行为数据集，对约 **1225 万条**（去重后 **621 万条**）用户行为记录进行分析。数据涵盖 2014 年 11 月 18 日至 12 月 18 日共 31 天，覆盖 **10,000 名用户**在淘宝平台上的点击、收藏、加购和购买行为，时间跨度包含 **双十二（12.12）购物节**。

### 数据说明

| 字段 | 说明 |
|------|------|
| user_id | 用户 ID（共 10,000 名） |
| item_id | 商品 ID（共 2,876,947 个） |
| behavior_type | 行为类型（1=点击, 2=收藏, 3=加购, 4=购买） |
| item_category | 商品类目 ID（共 8,916 个） |
| time | 行为发生时间（精确到小时） |

### 关键指标介绍

| 指标 | 含义 | 计算方式 |
|------|------|----------|
| PV（页面浏览量） | 用户总浏览次数 | 所有行为记录数 |
| UV（独立访客数） | 去重后的用户数 | 不重复的 user_id 数 |
| 转化率 | 从一个行为阶段到下一个阶段的比例 | 下一阶段用户数 / 上一阶段用户数 |
| 复购率 | 多次购买用户占购买用户的比例 | 购买天数 > 1 的用户数 / 总购买用户数 |
| 留存率 | 用户在某天活跃后，后续某天仍活跃的比例 | 第 N 天仍活跃用户数 / 基准日活跃用户数 |
| RFM 模型 | 用户价值评估框架 | R=最近购买距今天数, F=购买频次, M=购买商品种类数 |

## 分析目的

1. **了解用户行为模式**：分析点击、收藏、加购、购买四种行为的分布规律和时间趋势
2. **评估转化效率**：构建用户行为漏斗，量化各环节转化率，发现流失瓶颈
3. **挖掘用户价值**：基于 RFM 模型进行用户分群，识别高价值用户
4. **分析商品与类目表现**：找出热门类目和高转化类目，为选品和运营提供依据
5. **衡量用户粘性**：通过留存率和复购率评估平台对用户的吸引力
6. **评估双十二大促效果**：对比大促前后数据变化，衡量活动对用户行为的影响

## 结论与建议

**1. 购买行为占比极低，但双十二大促显著拉动成交**
> 数据：点击占 89.10%（553 万次），购买仅占 1.72%（10.7 万次）。但双十二当日购买量达 13,866 次，是日均购买量（3,441 次）的 **4.03 倍**。

> 建议：日常运营中转化率有较大提升空间。可借鉴双十二大促的"限时折扣 + 满减"模式，在非大促期间设置周期性小型促销（如每周特价日），持续刺激购买转化。

**2. 双十二大促吸引大量浏览但单日转化率反而下降**
> 数据：双十二当日 PV 达 346,464（日均 200,432 的 **1.73 倍**），但当日点击→购买转化率仅 **50.49%**，远低于非大促期间的 **87.34%**。

> 建议：大促期间涌入大量"逛而不买"的流量。应优化大促页面的商品推荐精准度，缩短用户决策路径；同时对大促期间收藏/加购但未购买的用户，在活动结束后 1~2 天内进行定向推送召回。

**3. 用户活跃高峰集中在晚间，符合大促倒计时场景**
> 数据：22 时为全天最高峰（552,227 次），20~22 时三个小时贡献了全天 **25.32%** 的行为量（157 万次）。

> 建议：将核心营销资源（秒杀、优惠券发放、直播带货）集中在 20:00~23:00 时段。双十二等大促的预热活动应在 20:00 前完成预告，22:00 推出限时爆款以承接流量高峰。

**4. 收藏/加购→购买转化率高，是大促蓄水的关键路径**
> 数据：93.24% 的用户有收藏或加购行为（9,324 人），其中 **95.30%** 最终完成了购买。而直接从点击到购买的整体转化率为 88.86%。

> 建议：收藏/加购是高意向行为，大促前应通过"提前加购享额外折扣"等机制引导用户蓄水。对加购未购买的用户，通过降价提醒、库存紧张通知等方式促进转化。

**5. 复购率高但复购间隔短，大促后需快速二次触达**
> 数据：87.17% 的购买用户（7,746 人）有复购行为，平均复购间隔仅 **3.7 天**，中位数为 **2 天**。

> 建议：用户购买后的 2~3 天是复购黄金窗口期。应在首次购买后及时推送关联商品推荐和复购优惠券。双十二结束后立即启动"返场活动"，抓住复购高峰。针对 RFM 模型中的"重要发展用户"制定专属复购激励策略。

**6. 热门点击类目与高转化类目存在错位**
> 数据：部分高点击量类目的购买转化率偏低，而一些中等流量类目反而拥有更高转化率。

> 建议：对高点击低转化类目进行诊断（价格偏高？详情页不足？竞品分流？），针对性优化。对高转化类目加大流量倾斜和广告投放，提升整体 GMV。双十二选品应优先纳入高转化类目作为引流主力。

```python
user_df = pd.read_csv('user_action.csv')
```

## 1. 数据预处理

```python
# 基本信息
print(f'数据形状: {user_df.shape}')
print(f'\n缺失值:\n{user_df.isnull().sum()}')
print(f'\n重复行数: {user_df.duplicated().sum()}')
print(f'\n数据类型:\n{user_df.dtypes}')
```

```python
# 去除重复值
user_df.drop_duplicates(inplace=True)

# 解析时间列，拆分日期和小时
user_df['date'] = user_df['time'].str[:10]
user_df['hour'] = user_df['time'].str[11:].astype(int)
user_df['date'] = pd.to_datetime(user_df['date'])

# 行为类型映射
behavior_map = {1: '点击', 2: '收藏', 3: '加购', 4: '购买'}
user_df['behavior'] = user_df['behavior_type'].map(behavior_map)

# 基础统计
print(f'用户数: {user_df["user_id"].nunique():,}')
print(f'商品数: {user_df["item_id"].nunique():,}')
print(f'类目数: {user_df["item_category"].nunique():,}')
print(f'时间范围: {user_df["date"].min().date()} ~ {user_df["date"].max().date()}')
print(f'去重后记录数: {len(user_df):,}')
user_df.head()
```

## 2. 用户行为总体分析

```python
# 各行为类型数量与占比
behavior_counts = user_df['behavior'].value_counts()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 柱状图
colors = ['#4C72B0', '#55A868', '#C44E52', '#8172B2']
behavior_counts.plot(kind='bar', ax=axes[0], color=colors)
axes[0].set_title('各行为类型数量', fontsize=14)
axes[0].set_xlabel('')
axes[0].set_ylabel('次数')
axes[0].tick_params(axis='x', rotation=0)
for i, v in enumerate(behavior_counts.values):
    axes[0].text(i, v + v*0.01, f'{v:,}', ha='center', fontsize=9)

# 饼图
axes[1].pie(behavior_counts.values, labels=behavior_counts.index, autopct='%1.2f%%',
            colors=colors, startangle=90)
axes[1].set_title('各行为类型占比', fontsize=14)

plt.tight_layout()
plt.show()
```

```python
# 每日 PV（页面浏览量）和 UV（独立访客数）趋势
daily_pv = user_df.groupby('date').size().rename('PV')
daily_uv = user_df.groupby('date')['user_id'].nunique().rename('UV')

fig, ax1 = plt.subplots(figsize=(14, 5))
ax2 = ax1.twinx()

ax1.plot(daily_pv.index, daily_pv.values, 'o-', color='#4C72B0', label='PV', linewidth=2)
ax2.plot(daily_uv.index, daily_uv.values, 's-', color='#C44E52', label='UV', linewidth=2)

ax1.set_xlabel('日期')
ax1.set_ylabel('PV（浏览量）', color='#4C72B0')
ax2.set_ylabel('UV（独立访客）', color='#C44E52')
ax1.tick_params(axis='x', rotation=45)
ax1.yaxis.set_major_formatter(ticker.FuncFormatter(lambda x, _: f'{x/1e4:.0f}万'))
ax2.yaxis.set_major_formatter(ticker.FuncFormatter(lambda x, _: f'{x/1e4:.1f}万'))

lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper left')

plt.title('每日 PV / UV 趋势', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
# 每小时行为分布 — 分析用户活跃时段
hourly = user_df.groupby(['hour', 'behavior']).size().unstack(fill_value=0)
hourly_order = ['点击', '收藏', '加购', '购买']
hourly = hourly[[c for c in hourly_order if c in hourly.columns]]

fig, axes = plt.subplots(1, 2, figsize=(16, 5))

# 总量
hourly_total = user_df.groupby('hour').size()
axes[0].bar(hourly_total.index, hourly_total.values, color='#4C72B0')
axes[0].set_title('每小时总行为量', fontsize=14)
axes[0].set_xlabel('小时')
axes[0].set_ylabel('行为次数')
axes[0].set_xticks(range(0, 24))

# 分行为
for col in hourly.columns:
    axes[1].plot(hourly.index, hourly[col], 'o-', label=col, linewidth=2)
axes[1].set_title('每小时各行为类型分布', fontsize=14)
axes[1].set_xlabel('小时')
axes[1].set_ylabel('行为次数')
axes[1].set_xticks(range(0, 24))
axes[1].legend()

plt.tight_layout()
plt.show()
```

### 2.3 双十二大促效果对比分析

对比大促前（12.01-12.11）、大促当日（12.12）、大促后（12.13-12.18）三个时段的核心指标变化，评估双十二活动对用户行为的拉动效果。

```python
# === 双十二大促效果对比分析 ===
# 定义三个时段
pre_promo = user_df[(user_df['date'] >= '2014-12-01') & (user_df['date'] <= '2014-12-11')]
on_promo = user_df[user_df['date'] == '2014-12-12']
post_promo = user_df[(user_df['date'] >= '2014-12-13') & (user_df['date'] <= '2014-12-18')]

# 计算各时段日均指标
def calc_period_metrics(df, days):
    """计算某时段的日均 PV、UV、购买量"""
    pv = len(df) / days
    uv = df.groupby('date')['user_id'].nunique().mean()
    purchases = len(df[df['behavior_type'] == 4]) / days
    return pv, uv, purchases

pre_pv, pre_uv, pre_buy = calc_period_metrics(pre_promo, 11)
on_pv, on_uv, on_buy = calc_period_metrics(on_promo, 1)
post_pv, post_uv, post_buy = calc_period_metrics(post_promo, 6)

# 打印对比数据
print('=' * 60)
print('双十二大促效果对比（日均指标）')
print('=' * 60)
print(f'{"指标":<12} {"大促前(12.01-11)":<18} {"双十二当日":<18} {"大促后(12.13-18)":<18}')
print('-' * 60)
print(f'{"日均PV":<12} {pre_pv:>14,.0f}   {on_pv:>14,.0f}   {post_pv:>14,.0f}')
print(f'{"日均UV":<12} {pre_uv:>14,.0f}   {on_uv:>14,.0f}   {post_uv:>14,.0f}')
print(f'{"日均购买量":<10} {pre_buy:>14,.0f}   {on_buy:>14,.0f}   {post_buy:>14,.0f}')

# 计算提升幅度（相对大促前）
print(f'\n--- 双十二当日 vs 大促前 提升幅度 ---')
print(f'PV 提升: {(on_pv / pre_pv - 1) * 100:+.1f}%')
print(f'UV 提升: {(on_uv / pre_uv - 1) * 100:+.1f}%')
print(f'购买量提升: {(on_buy / pre_buy - 1) * 100:+.1f}%')

print(f'\n--- 大促后 vs 大促前 变化 ---')
print(f'PV 变化: {(post_pv / pre_pv - 1) * 100:+.1f}%')
print(f'UV 变化: {(post_uv / pre_uv - 1) * 100:+.1f}%')
print(f'购买量变化: {(post_buy / pre_buy - 1) * 100:+.1f}%')

# 分组柱状图
metrics = ['日均PV', '日均UV', '日均购买量']
pre_vals = [pre_pv, pre_uv, pre_buy]
on_vals = [on_pv, on_uv, on_buy]
post_vals = [post_pv, post_uv, post_buy]

fig, axes = plt.subplots(1, 3, figsize=(16, 5))
x = np.arange(3)
period_labels = ['大促前\n(12.01-12.11)', '双十二\n(12.12)', '大促后\n(12.13-12.18)']
colors_period = ['#4C72B0', '#C44E52', '#55A868']

for idx, (metric, pre_v, on_v, post_v) in enumerate(zip(metrics, pre_vals, on_vals, post_vals)):
    vals = [pre_v, on_v, post_v]
    bars = axes[idx].bar(period_labels, vals, color=colors_period, width=0.5)
    axes[idx].set_title(metric, fontsize=13)
    for bar, v in zip(bars, vals):
        axes[idx].text(bar.get_x() + bar.get_width()/2, v + max(vals)*0.02,
                      f'{v:,.0f}', ha='center', fontsize=10)
    # 标注提升率
    uplift = (on_v / pre_v - 1) * 100
    axes[idx].annotate(f'{uplift:+.1f}%', xy=(1, on_v), xytext=(1.3, on_v * 0.9),
                      fontsize=10, color='#C44E52', fontweight='bold',
                      arrowprops=dict(arrowstyle='->', color='#C44E52'))

plt.suptitle('双十二大促效果对比（日均指标）', fontsize=15, y=1.02)
plt.tight_layout()
plt.show()
```

## 3. 用户转化漏斗分析

```python
# 转化漏斗：按独立用户数统计
click_users = user_df[user_df['behavior_type'] == 1]['user_id'].nunique()
fav_users = user_df[user_df['behavior_type'] == 2]['user_id'].nunique()
cart_users = user_df[user_df['behavior_type'] == 3]['user_id'].nunique()
buy_users = user_df[user_df['behavior_type'] == 4]['user_id'].nunique()
fav_cart_users = user_df[user_df['behavior_type'].isin([2, 3])]['user_id'].nunique()

funnel_labels = ['点击', '收藏/加购', '购买']
funnel_values = [click_users, fav_cart_users, buy_users]
funnel_rates = [v / funnel_values[0] * 100 for v in funnel_values]

fig, ax = plt.subplots(figsize=(10, 6))
colors_funnel = ['#4C72B0', '#55A868', '#C44E52']

# 水平漏斗柱状图
y_pos = range(len(funnel_labels))
bars = ax.barh(y_pos, funnel_values, color=colors_funnel, height=0.5)
ax.set_yticks(y_pos)
ax.set_yticklabels(funnel_labels, fontsize=13)
ax.invert_yaxis()
ax.set_title('用户转化漏斗（独立用户数）', fontsize=14)

for i, (v, r) in enumerate(zip(funnel_values, funnel_rates)):
    ax.text(v + click_users * 0.01, i, f'{v:,} 人 ({r:.2f}%)', va='center', fontsize=11)

plt.tight_layout()
plt.show()

# 逐级转化率
print(f'点击 → 收藏/加购 转化率: {fav_cart_users/click_users*100:.2f}%')
print(f'收藏/加购 → 购买 转化率: {buy_users/fav_cart_users*100:.2f}%')
print(f'点击 → 购买 总转化率: {buy_users/click_users*100:.2f}%')
```

```python
# === 分时段转化漏斗对比：大促前 vs 双十二 vs 大促后 ===
# 展示大促对各环节转化率的影响，体现指标体系搭建能力

def calc_funnel(df, period_name):
    """计算某时段的转化漏斗（独立用户数 & 转化率）"""
    click_u = df[df['behavior_type'] == 1]['user_id'].nunique()
    fav_cart_u = df[df['behavior_type'].isin([2, 3])]['user_id'].nunique()
    buy_u = df[df['behavior_type'] == 4]['user_id'].nunique()
    return {
        '时段': period_name,
        '点击用户数': click_u,
        '收藏/加购用户数': fav_cart_u,
        '购买用户数': buy_u,
        '点击→收藏加购率': fav_cart_u / click_u * 100 if click_u else 0,
        '收藏加购→购买率': buy_u / fav_cart_u * 100 if fav_cart_u else 0,
        '整体转化率': buy_u / click_u * 100 if click_u else 0,
    }

# 使用前面已定义的 pre_promo, on_promo, post_promo
funnel_pre = calc_funnel(pre_promo, '大促前(12.01-11)')
funnel_on = calc_funnel(on_promo, '双十二(12.12)')
funnel_post = calc_funnel(post_promo, '大促后(12.13-18)')

funnel_df = pd.DataFrame([funnel_pre, funnel_on, funnel_post])
print('=== 分时段转化漏斗对比 ===')
print(funnel_df.to_string(index=False))

# 分组柱状图：对比三个时段各环节转化率
fig, ax = plt.subplots(figsize=(12, 6))
rate_cols = ['点击→收藏加购率', '收藏加购→购买率', '整体转化率']
x = np.arange(len(rate_cols))
width = 0.25

bars1 = ax.bar(x - width, funnel_df[rate_cols].iloc[0].values, width, label='大促前(12.01-11)', color='#4C72B0')
bars2 = ax.bar(x, funnel_df[rate_cols].iloc[1].values, width, label='双十二(12.12)', color='#C44E52')
bars3 = ax.bar(x + width, funnel_df[rate_cols].iloc[2].values, width, label='大促后(12.13-18)', color='#55A868')

ax.set_xticks(x)
ax.set_xticklabels(rate_cols, fontsize=12)
ax.set_ylabel('转化率 (%)')
ax.set_title('分时段转化漏斗对比（大促前 / 双十二 / 大促后）', fontsize=14)
ax.legend(fontsize=11)

# 标注数值
for bars in [bars1, bars2, bars3]:
    for bar in bars:
        h = bar.get_height()
        ax.text(bar.get_x() + bar.get_width()/2, h + 0.5, f'{h:.1f}%', ha='center', fontsize=9)

plt.tight_layout()
plt.show()

# 分析洞察
print(f'\n--- 洞察 ---')
print(f'双十二当日整体转化率: {funnel_on["整体转化率"]:.2f}%，大促前: {funnel_pre["整体转化率"]:.2f}%')
print(f'转化率变化: {funnel_on["整体转化率"] - funnel_pre["整体转化率"]:+.2f} 个百分点')
print(f'说明：大促期间虽然流量大幅增加，但需关注转化效率的变化，为后续大促优化提供依据。')
```

## 4. 用户价值分析（RFM 模型）

```python
# RFM 分析（基于购买行为）
buy_df = user_df[user_df['behavior_type'] == 4].copy()
max_date = user_df['date'].max()

rfm = buy_df.groupby('user_id').agg(
    R=('date', lambda x: (max_date - x.max()).days),
    F=('item_id', 'count'),
    M=('item_id', 'nunique')  # 无金额数据，用购买商品种类数替代
).reset_index()

# 打分（分为 1~4 分，越高越好）
rfm['R_score'] = pd.qcut(rfm['R'], 4, labels=[4, 3, 2, 1], duplicates='drop')
rfm['F_score'] = pd.qcut(rfm['F'].rank(method='first'), 4, labels=[1, 2, 3, 4], duplicates='drop')
rfm['M_score'] = pd.qcut(rfm['M'].rank(method='first'), 4, labels=[1, 2, 3, 4], duplicates='drop')

# 用户分群：每个维度与中位数比较
rfm['R_label'] = (rfm['R_score'].astype(int) >= rfm['R_score'].astype(int).median()).astype(int)
rfm['F_label'] = (rfm['F_score'].astype(int) >= rfm['F_score'].astype(int).median()).astype(int)
rfm['M_label'] = (rfm['M_score'].astype(int) >= rfm['M_score'].astype(int).median()).astype(int)

def classify_user(row):
    r, f, m = row['R_label'], row['F_label'], row['M_label']
    if r and f and m: return '重要价值用户'
    elif r and f and not m: return '重要保持用户'
    elif r and not f and m: return '重要发展用户'
    elif r and not f and not m: return '重要挽留用户'
    elif not r and f and m: return '一般价值用户'
    elif not r and f and not m: return '一般保持用户'
    elif not r and not f and m: return '一般发展用户'
    else: return '一般挽留用户'

rfm['用户类型'] = rfm.apply(classify_user, axis=1)

# 可视化
user_type_counts = rfm['用户类型'].value_counts()

fig, ax = plt.subplots(figsize=(10, 6))
user_type_counts.plot(kind='barh', color='#4C72B0', ax=ax)
ax.set_title('RFM 用户分群分布', fontsize=14)
ax.set_xlabel('用户数')
for i, v in enumerate(user_type_counts.values):
    ax.text(v + user_type_counts.max() * 0.01, i, f'{v:,}', va='center')
plt.tight_layout()
plt.show()

print(f'\n有购买行为的用户总数: {len(rfm):,}')
print(rfm['用户类型'].value_counts())
```

```python
# === RFM 用户分群 × 双十二行为交叉分析 ===
# 分析不同 RFM 用户群体在双十二的行为表现，指导大促精细化运营

# 双十二当日行为数据
promo_day = user_df[user_df['date'] == '2014-12-12'].copy()

# 将 RFM 分群结果合并到双十二数据
promo_rfm = promo_day.merge(rfm[['user_id', '用户类型']], on='user_id', how='inner')

# 各用户群体在双十二的行为统计
rfm_promo_stats = promo_rfm.groupby(['用户类型', 'behavior']).size().unstack(fill_value=0)
# 确保列顺序一致
for col in ['点击', '收藏', '加购', '购买']:
    if col not in rfm_promo_stats.columns:
        rfm_promo_stats[col] = 0
rfm_promo_stats = rfm_promo_stats[['点击', '收藏', '加购', '购买']]

print('=== 各 RFM 用户群体在双十二当日的行为量 ===')
print(rfm_promo_stats)

# 计算各群体双十二购买转化率（购买用户数 / 活跃用户数）
rfm_promo_users = promo_rfm.groupby('用户类型').agg(
    活跃用户数=('user_id', 'nunique'),
    购买用户数=('user_id', lambda x: x[promo_rfm.loc[x.index, 'behavior_type'] == 4].nunique()),
    总行为量=('user_id', 'count')
)
rfm_promo_users['双十二转化率'] = rfm_promo_users['购买用户数'] / rfm_promo_users['活跃用户数'] * 100
rfm_promo_users = rfm_promo_users.sort_values('双十二转化率', ascending=False)

print(f'\n=== 各 RFM 群体双十二转化率 ===')
print(rfm_promo_users)

# 热力图可视化
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# 左图：各群体行为量热力图
import matplotlib.colors as mcolors
rfm_promo_norm = rfm_promo_stats.div(rfm_promo_stats.sum(axis=1), axis=0) * 100
im = axes[0].imshow(rfm_promo_norm.values, cmap='YlOrRd', aspect='auto')
axes[0].set_xticks(range(len(rfm_promo_norm.columns)))
axes[0].set_xticklabels(rfm_promo_norm.columns, fontsize=11)
axes[0].set_yticks(range(len(rfm_promo_norm.index)))
axes[0].set_yticklabels(rfm_promo_norm.index, fontsize=10)
axes[0].set_title('各 RFM 群体双十二行为占比 (%)', fontsize=13)
for i in range(len(rfm_promo_norm.index)):
    for j in range(len(rfm_promo_norm.columns)):
        axes[0].text(j, i, f'{rfm_promo_norm.values[i, j]:.1f}%',
                    ha='center', va='center', fontsize=9,
                    color='white' if rfm_promo_norm.values[i, j] > 50 else 'black')
plt.colorbar(im, ax=axes[0], shrink=0.8)

# 右图：各群体双十二转化率
colors_bar = ['#C44E52' if v > rfm_promo_users['双十二转化率'].median() else '#4C72B0'
              for v in rfm_promo_users['双十二转化率'].values]
axes[1].barh(rfm_promo_users.index, rfm_promo_users['双十二转化率'], color=colors_bar)
axes[1].set_title('各 RFM 群体双十二购买转化率', fontsize=13)
axes[1].set_xlabel('转化率 (%)')
for i, v in enumerate(rfm_promo_users['双十二转化率'].values):
    axes[1].text(v + 0.5, i, f'{v:.1f}%', va='center', fontsize=10)

plt.suptitle('RFM 用户分群 × 双十二大促行为分析', fontsize=15, y=1.02)
plt.tight_layout()
plt.show()

print(f'\n--- 运营建议 ---')
print(f'转化率最高群体: {rfm_promo_users.index[0]}（{rfm_promo_users["双十二转化率"].iloc[0]:.1f}%）')
print(f'转化率最低群体: {rfm_promo_users.index[-1]}（{rfm_promo_users["双十二转化率"].iloc[-1]:.1f}%）')
print(f'建议：针对高转化群体加大大促触达力度，针对低转化群体设计专属优惠券提升参与度。')
```

```python
# === 热门类目用户行为路径分析 ===
# 分析 Top 类目中用户从点击到购买的完整路径转化率，体现用户行为路径分析能力

# 选取购买量 Top 10 类目
top_cats = user_df[user_df['behavior_type'] == 4]['item_category'].value_counts().head(10).index.tolist()
path_df = user_df[user_df['item_category'].isin(top_cats)].copy()

# 统计每个类目各行为的独立用户数
path_results = []
for cat in top_cats:
    cat_data = path_df[path_df['item_category'] == cat]
    click_u = cat_data[cat_data['behavior_type'] == 1]['user_id'].nunique()
    fav_u = cat_data[cat_data['behavior_type'] == 2]['user_id'].nunique()
    cart_u = cat_data[cat_data['behavior_type'] == 3]['user_id'].nunique()
    buy_u = cat_data[cat_data['behavior_type'] == 4]['user_id'].nunique()
    
    path_results.append({
        '类目ID': cat,
        '点击用户': click_u,
        '收藏用户': fav_u,
        '加购用户': cart_u,
        '购买用户': buy_u,
        '点击→收藏率': fav_u / click_u * 100 if click_u else 0,
        '点击→加购率': cart_u / click_u * 100 if click_u else 0,
        '点击→购买率': buy_u / click_u * 100 if click_u else 0,
        '收藏→购买率': buy_u / fav_u * 100 if fav_u else 0,
        '加购→购买率': buy_u / cart_u * 100 if cart_u else 0,
    })

path_table = pd.DataFrame(path_results)

# 打印路径转化率表
print('=== 热门类目（购买量 Top 10）用户行为路径转化率 ===')
display_cols = ['类目ID', '点击用户', '收藏用户', '加购用户', '购买用户',
                '点击→收藏率', '点击→加购率', '点击→购买率', '加购→购买率']
print(path_table[display_cols].to_string(index=False, float_format='%.2f'))

# 可视化：多类目漏斗对比
fig, axes = plt.subplots(1, 2, figsize=(16, 7))

# 左图：各类目各环节转化率对比
rate_cols_path = ['点击→收藏率', '点击→加购率', '点击→购买率']
x = np.arange(len(top_cats))
width = 0.25
colors_path = ['#4C72B0', '#55A868', '#C44E52']

for i, col in enumerate(rate_cols_path):
    axes[0].bar(x + i * width, path_table[col].values, width, label=col, color=colors_path[i])

axes[0].set_xticks(x + width)
axes[0].set_xticklabels(path_table['类目ID'].astype(str), rotation=45, fontsize=9)
axes[0].set_ylabel('转化率 (%)')
axes[0].set_title('Top 10 类目各环节转化率对比', fontsize=13)
axes[0].legend(fontsize=10)

# 右图：加购→购买转化率排名（衡量购买决策效率）
path_sorted = path_table.sort_values('加购→购买率', ascending=True)
colors_conv = ['#C44E52' if v > path_sorted['加购→购买率'].median() else '#4C72B0'
               for v in path_sorted['加购→购买率'].values]
axes[1].barh(path_sorted['类目ID'].astype(str), path_sorted['加购→购买率'], color=colors_conv)
axes[1].set_title('Top 10 类目：加购→购买转化率', fontsize=13)
axes[1].set_xlabel('转化率 (%)')
for i, v in enumerate(path_sorted['加购→购买率'].values):
    axes[1].text(v + 0.5, i, f'{v:.1f}%', va='center', fontsize=10)

plt.suptitle('热门类目用户行为路径分析', fontsize=15, y=1.02)
plt.tight_layout()
plt.show()

# 路径洞察
best_cat = path_table.loc[path_table['加购→购买率'].idxmax()]
worst_cat = path_table.loc[path_table['加购→购买率'].idxmin()]
print(f'\n--- 路径分析洞察 ---')
print(f'加购→购买转化率最高类目: {int(best_cat["类目ID"])}（{best_cat["加购→购买率"]:.1f}%），用户决策效率高')
print(f'加购→购买转化率最低类目: {int(worst_cat["类目ID"])}（{worst_cat["加购→购买率"]:.1f}%），存在犹豫流失')
print(f'建议：对低转化类目可通过限时优惠、库存提醒等方式缩短用户决策周期。')
```

## 6. 用户留存与复购分析

```python
# 日留存率分析（基于活跃用户）
dates = sorted(user_df['date'].unique())
retention_data = {}

for i, base_date in enumerate(dates[:-1]):
    base_users = set(user_df[user_df['date'] == base_date]['user_id'].unique())
    base_count = len(base_users)
    retention_data[base_date] = {}
    for j, future_date in enumerate(dates[i+1:], 1):
        future_users = set(user_df[user_df['date'] == future_date]['user_id'].unique())
        retained = len(base_users & future_users)
        retention_data[base_date][f'Day{j}'] = retained / base_count * 100 if base_count > 0 else 0

# 平均留存率
max_days = min(7, len(dates) - 1)
avg_retention = []
for d in range(1, max_days + 1):
    key = f'Day{d}'
    vals = [v.get(key, 0) for v in retention_data.values() if key in v]
    avg_retention.append(np.mean(vals) if vals else 0)

fig, ax = plt.subplots(figsize=(10, 5))
days_label = [f'Day{i}' for i in range(1, max_days + 1)]
ax.plot(days_label, avg_retention, 'o-', color='#4C72B0', linewidth=2, markersize=8)
ax.set_title('平均日留存率', fontsize=14)
ax.set_xlabel('留存天数')
ax.set_ylabel('留存率 (%)')
for i, v in enumerate(avg_retention):
    ax.text(i, v + 0.5, f'{v:.1f}%', ha='center', fontsize=10)
ax.set_ylim(0, max(avg_retention) * 1.2 if avg_retention else 100)
plt.tight_layout()
plt.show()
```

```python
# 复购率分析
buy_user_freq = buy_df.groupby('user_id')['date'].nunique()

total_buy_users = len(buy_user_freq)
repeat_buy_users = (buy_user_freq > 1).sum()
repeat_rate = repeat_buy_users / total_buy_users * 100

print(f'有购买行为的用户数: {total_buy_users:,}')
print(f'复购用户数（购买天数 > 1）: {repeat_buy_users:,}')
print(f'复购率: {repeat_rate:.2f}%')

# 购买频次分布
fig, ax = plt.subplots(figsize=(10, 5))
freq_dist = buy_user_freq.value_counts().sort_index().head(15)
ax.bar(freq_dist.index.astype(str), freq_dist.values, color='#4C72B0')
ax.set_title('用户购买天数分布', fontsize=14)
ax.set_xlabel('购买天数')
ax.set_ylabel('用户数')
for i, v in enumerate(freq_dist.values):
    ax.text(i, v + v*0.02, f'{v:,}', ha='center', fontsize=9)
plt.tight_layout()
plt.show()
```

```python
# 复购天数分布（复购用户两次购买之间的间隔天数）
repeat_users = buy_user_freq[buy_user_freq > 1].index
repeat_buy_df = buy_df[buy_df['user_id'].isin(repeat_users)].copy()

# 计算每个用户相邻两次购买的间隔天数
repeat_buy_df = repeat_buy_df.sort_values(['user_id', 'date'])
repeat_buy_df['prev_date'] = repeat_buy_df.groupby('user_id')['date'].shift(1)
repeat_buy_df['gap_days'] = (repeat_buy_df['date'] - repeat_buy_df['prev_date']).dt.days
gap_days = repeat_buy_df['gap_days'].dropna().astype(int)
gap_days = gap_days[gap_days > 0]

fig, ax = plt.subplots(figsize=(12, 5))
gap_dist = gap_days.value_counts().sort_index()
ax.bar(gap_dist.index.astype(str), gap_dist.values, color='#8172B2')
ax.set_title('复购天数间隔分布', fontsize=14)
ax.set_xlabel('间隔天数')
ax.set_ylabel('次数')
for i, v in enumerate(gap_dist.values):
    ax.text(i, v + v * 0.02, f'{v:,}', ha='center', fontsize=9)
plt.tight_layout()
plt.show()

print(f'平均复购间隔: {gap_days.mean():.1f} 天')
print(f'中位数复购间隔: {gap_days.median():.0f} 天')
```
